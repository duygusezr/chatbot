```python
from transformers import pipeline

# Türkçe SQuAD modeli ile QA pipeline'ı oluştur
qa_pipeline = pipeline(
    "question-answering",
    model="savasy/bert-base-turkish-squad",
    tokenizer="savasy/bert-base-turkish-squad"
)

context = """
Film, New York’un karanlık arka sokaklarında yaşayan yalnız bir suikastçı olan Leon’un hikayesini konu alır. Leon, profesyonel bir “temizleyici”dir; görevlerini sessiz, dikkatli ve duygusuz biçimde yerine getirir. Hayatındaki tek düzenli şey, her sabah içtiği süt ve büyük bir özenle baktığı bitkisidir. Bu bitki, onun için özgürlük ve kök salamamanın bir sembolüdür. Leon’un geçmişi tam olarak açıklanmasa da, travmalarla dolu olduğu ve bu yüzden yalnızlığı seçtiği hissedilir.

Hikâyeye 12 yaşındaki Matilda’nın girmesiyle her şey değişir. Matilda, aynı apartmanda yaşayan, ilgisiz bir aileye sahip, sigara içen ve yaşıtlarından farklı bir çocuktur. Ailesi, yozlaşmış ve acımasız bir DEA ajanı olan Stansfield ve adamları tarafından vahşice öldürülür. O sırada alışverişte olan Matilda, eve döndüğünde büyük bir korkuyla Leon’un kapısını çalar. Leon, ilk başta kararsızlık yaşasa da onu içeri alarak hayatını kurtarır. Bu karar, Leon’un hayatında ilk kez bir başkasına gerçekten bağlanmasına neden olur.

Zamanla aralarında güçlü bir bağ gelişir. Matilda, ailesinin intikamını almak ister ve Leon’dan kendisine silah kullanmayı öğretmesini ister. Leon, başlangıçta bu fikre karşı çıkar ancak sonunda ona hedef alma, hayatta kalma ve sessizce iz sürme gibi becerileri öğretir. Bu süreçte Matilda, Leon’un hayatına anlam katarken, Leon da Matilda’ya güven, aidiyet ve korunma duygusu verir. Aralarındaki ilişki asla romantik değildir; daha çok bir baba-kız bağı gibi koruyucu ve duygusal bir ilişkidir.

Film boyunca Matilda’nın karakter gelişimi dikkat çekicidir. Başlarda kırılgan, öfkeli ve yalnız bir çocuk olan Matilda, Leon sayesinde güçlü, kararlı ve ne istediğini bilen bir bireye dönüşür. Bu süreç, izleyiciye sevgi, güven, fedakarlık ve masumiyetin korunması gibi evrensel temaları etkileyici bir şekilde aktarır.

Filmin zirve noktasında, Stansfield Leon’un yerini öğrenir ve büyük bir baskın gerçekleştirir. Leon, Matilda’yı kurtarabilmek için kendini feda eder. Ölmeden önce Matilda’yı dışarı çıkarır ve onun hayatta kalmasını sağlar. Son sözleri, onun artık sadece bir katil değil; koruyucu, bağlı ve fedakar bir insan olduğunu gösterir: “Onu hayatta tut. O, benden daha çok yaşamaya değer.”

Film, Matilda’nın Leon’un çok sevdiği bitkiyi bir okul bahçesine ekmesiyle sona erer. Bu sahne, onun Leon sayesinde kök salmaya başladığını, artık hayata umutla bakabildiğini simgeler. “Leon: Sevginin Gücü”, bir çocuğun masumiyetiyle bir katilin yalnız kalbini değiştirdiği, karanlıkla aydınlığın kesiştiği etkileyici bir hikayedir.
"""

print("Soru-cevap moduna geçildi. Çıkmak için 'çık' yazın.")
while True:
    question = input("Soru: ")
    if question.lower() == "çık":
        break
    result = qa_pipeline(question=question, context=context)
    # Eğer modelin cevabı çok kısa veya alakasızsa "Bilmiyorum" diyebilirsin
    if result['score'] < 0.2 or result['answer'].strip() == "":
        print("Cevap: Bilmiyorum.")
    else:
        print("Cevap:", result['answer'])
```
